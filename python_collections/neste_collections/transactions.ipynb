```python
#Total amount through UPI.

total_upi_payment = sum(t["amount"] for t in transactions  if t["transaction_method"]=="UPI")
total_upi_payment
```

```python
#Average amount by method

methods = {t["transaction_method"] for t in transactions}

average_by_method = {
    m: sum(t["amount"] for t in transactions if t["transaction_method"] == m)
       / len([t for t in transactions if t["transaction_method"] == m])
    for m in methods
}
average_by_method
```
