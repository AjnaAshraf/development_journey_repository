### movie casestudy

```python
movies=[
            ["kgf","kannada",150,2023,8],
            ["balan","malayalam",160,2026,7],
            ["athiradi","malayalam",180,2025,8],
            ["ramayan","hindi",160,2026,9],
            ["goatlife","malayalam",180,2026,8],
            ["spiderman:brand new day","english",180,2027,9]
]
```

```python
#all movie titles

all_movie_titles=[m[0] for m in movies]
all_movie_titles
```

```python
all_languages={m[1] for m in movies}
all_languages
```

```python
all_movie_title_upper=[m[0].upper() for m in movies]
all_movie_title_upper
```

```python
# movie title starting with a 
movie_titile_starting_a =[m[0] for m in movies if m[0].startswith("a")]
movie_titile_starting_a
```

```python
# movie title starting with k
movie_titile_starting_a =[m[0] for m in movies if m[0][0]=="k"]
movie_titile_starting_a
```

```python
#display all malayalam movies
malayalam_movies=[m[0] for m in movies if m[1]=="malayalam"]
malayalam_movies
```

```python
language=[m[1] for m in movies]
lc={l:language.count(l) for l in language}
lc
```

```python
import matplotlib.pyplot as plt

x=lc.keys()
y=lc.values()
plt.bar(x,y)
```

```python
#highest runtime movie
highest_runtime = max(m[2] for m in movies)
highest_runtime_movie = [m[0] for m in movies if m[2] == highest_runtime]
highest_runtime_movie
```

```python
#lowest runtime movie

lowest_runtime = min(m[2] for m in movies)
lowest_runtime_movie = [m[0] for m in movies if m[2] == lowest_runtime]
lowest_runtime_movie
```

```python
#avg runtime
runtime =[m[2] for m in movies]
avg=sum(runtime)//len(runtime)
avg
```

```python
yr = [m[3] for m in movies]
yr_wise = {y:yr.count(y) for y in yr}
yr_wise
```

```python
x=yr_wise.keys()
y=yr_wise.values()
plt.bar(x,y)
```
