### movie case study 2

```python
movies_data = [
["dangal", "hindi", 161, 2016, 8.3], # 0
["premam", "malayalam", 156, 2015, 8.3], # 1
["inception", "english", 148, 2010, 8.8], # 2
["vikram", "tamil", 175, 2022, 8.3], # 3
["parasite", "korean", 132, 2019, 8.5], # 4
["rrr", "telugu", 187, 2022, 7.8], # 5
["drishyam", "malayalam", 160, 2013, 8.3], # 6
["avatar", "english", 162, 2009, 7.9], # 7
["kumbalangi nights", "malayalam", 135, 2019, 8.5], # 8
["sholay", "hindi", 204, 1975, 8.1] # 9
]
```

### basic filtering

```python
#Find by Language: Print the titles of all Malayalam movie

language = [m[0] for m in movies_data if m[1]=="malayalam"]
language
```

```python
#Find by Year: Print the titles of all movies released after the year 2015.

movies=[m[0] for m in movies_data if m[3]>2015]
movies
```

```python
#Find by Rating: Count how many movies have a rating of 8.5 or higher.

rating = [m for m in movies_data if m[4]>=8.5]
len(rating)
```

### intermediate filtering

```python
#Average Duration: Calculate the average duration (in minutes) of all the movies in the dataset.

duration =sum([m[2] for m in movies_data])
avg=duration/len(movies_data)
avg
```

```python
#Q5. Multiple Conditions: Print the titles of movies that are longer than 150 minutes AND have a rating greater than 8.0.

new_mov=[m[0] for m in movies_data if m[2]>150 and m[4]>8.0]
new_mov
```

### Advanced searching

```python
# Q6. Highest Rating: Find and print the entire record for the movie with the highest rating

rating = max([m[4]for m in movies_data])
highest_rating =[m for m in movies_data if m[4]==rating]
highest_rating
```

```python
#Q7. Oldest Movie: Find the title and release year of the oldest movie in the dataset
min_yr =min([m[3] for m in movies_data])
movie=[m[0]  for m in movies_data if m[3]== min_yr]
movie,min_yr
```
