### LAMBDA FUNCTION
DEFINITION : Anonymous function with a single expression

### syntax : lambda p1,p2 : expression

eg:

add_num = lambda n1,n2 : n1+n2


```python
cube = lambda n : n**3
cube(3)
```

```python
#adding to numbers

add_numbers = lambda n1,n2 : n1+n2
add_numbers(100,200)
```

```python
#subtracting two number

sub_num = lambda n1,n2 : n1-n2
sub_num(200,100)
```

```python
#max  of two numbers

max_two = lambda n1,n2 : n1 if n1>n2 else n2

max_two(10,20)
```

```python
max_num = lambda n1,n2 : max(n1,n2)
max_num(100,200)
```

```python
min_two = lambda n1,n2: n1 if n1<n2 else n2
min_two(200,201)
```

```python
user_experience = [
                    ["hari",6],
                    ["arun",2],
                    ["vinay",4],
                    ["hana",1],
                    ["ajna",6],
                    ["anjali",2]
]

max(user_experience)
```

```python
print("maximum experience ",max(user_experience,key=lambda lst : lst[1]))
print("minimum experience ",min(user_experience,key=lambda lst : lst[1]))
```

```python
weather= [

            {"district":"tvm","temp":25},
            {"district":"kty","temp":26},
            {"district":"ekm","temp":37},
            {"district":"alp","temp":29},
            {"district":"pkd","temp":42},
            {"district":"tsr","temp":36}
]

print("maximum temperature: ",max(weather,key=lambda dictionary : dictionary.get("temp")))
print("minimum temperature: ",min(weather,key=lambda dictionary : dictionary.get("temp")))
```
