 1) Movie title,language,year,director,genre 

    -setmovie(title,language,year,director,genre)-getmovie(self)

```python
class Movie:

    title:str
    language:str
    year:int
    director:str
    genre:str

    def __init__(self,title,language,year,director,genre):

        self.title=title
        self.language=language
        self.year=year
        self.director=director
        self.genre=genre

    def get_Movie(self):

        print(self.title,self.language,self.year,self.director,self.genre)

project_hail_mary = Movie("Project Hail Mary,","English,",2026," ,Phil Lord and Christopher Miller",",Sci-Fi")
super_man = Movie("Superman,","English,",2025,",James Gunn,","DC -Adventure,superhero")
The_Shawshank_Redemption = Movie("The Shawshank Redemption,","English,",1994,",Frank Darabont",",Psychological Drama")

project_hail_mary.get_Movie()
super_man.get_Movie()
The_Shawshank_Redemption.get_Movie()
```

        -setemployee(id,name,salary,phone,department)

```python
class Employee:

    employee_id :str
    name:str
    salary:float
    phone:int
    department:str

    def __init__(self,employee_id,name,salary,phone,department):
        self.employee_id=employee_id
        self.name=name
        self.salary=salary
        self.phone=phone
        self.department=department

    def get_employee(self):

        print(self.employee_id,self.name,self.salary,self.phone,self.department)

ajna=Employee("CN01","Ajna Ashraf",45000,9747370000,"IT")
sana=Employee("MB11","Sana Fathima",36000,5652928627,"Hr")

ajna.get_employee()
sana.get_employee()
```

### Apps

```python
class Apps:

    App_name:str
    category:str
    launch_year:int
    founder:str

    def __init__(self,App_name,category,launch_year,founder):

        self.App_name=App_name
        self.category=category
        self.launch_year=launch_year
        self.founder=founder

    def get_App(self):

        print(self.App_name,self.category,self.launch_year,self.founder)

instagram = Apps("Instagram","Entertainment-photos and videos",2010,"Kevin Systrom and Mike Krieger")
instagram.get_App()

swiggy = Apps("Swiggy","Delivery -foods and drinks",2014," Sriharsha Majety, Nandan Reddy, and Rahul Jaimini")
swiggy.get_App()

Whatsapp = Apps("Whatsapp","Social networking",2009," Jan Koum and Brian Acton")
Whatsapp.get_App()

```
